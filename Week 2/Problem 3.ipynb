```python
import numpy as np
```

```python
# Given Data
x = np.array([1.0, 1.1, 1.3, 1.5, 1.9, 2.1])
y = np.array([1.84, 1.96, 2.21, 2.45, 2.94, 3.18])
```

```python
# Utilize the singular value decomposition method to find the linear least squares polynomial of degree 2

X = np.vstack([np.ones_like(x), x, x**2]).T


U, S, Vt = np.linalg.svd(X, full_matrices=False)
S_inv = np.diag(1 / S)


b = Vt.T @ S_inv @ U.T @ y

print("Coefficients (a0, a1, a2):", b)
```

```python
# Utilize the singular value decomposition method to find the linear least squares polynomial of degree 3

X = np.vstack([np.ones_like(x), x, x**2, x**3]).T


U, S, Vt = np.linalg.svd(X, full_matrices=False)
S_inv = np.diag(1 / S)


b = Vt.T @ S_inv @ U.T @ y

print("Coefficients (a0, a1, a2, a3):", b)
```
